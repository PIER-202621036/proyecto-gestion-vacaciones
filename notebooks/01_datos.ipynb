# **Sistema de gestión y análisis de vacaciones**

**1. Problema a resolver**

La empresa objeto de estudio presenta una deficiente gestión de vacaciones la cual genera acumulación de descansos, costos no planificados y poca visibilidad para priorizar salidas. Para resolverlo, el proyecto busca equilibrar la atención de vacaciones con la criticidad de los puestos, protegiendo la continuidad operativa de la empresa.

---

**2. Solución propuesta**

Desarrollar progresivamente un sistema de gestión y análisis de vacaciones. Durante las primeras siete semanas del curso, el equipo establecerá los fundamentos del proyecto utilizando Python, Google Colab, Excel, NumPy y Pandas para identificar los riesgos de acumulación, evaluar las prioridades de atención y estimar el impacto económico de las solicitudes de los empleados. La solución integrará reglas de negocio, validaciones y visualizaciones, complementándose posteriormente con bases de datos SQL e interfaz gráfica para la toma de decisiones en Recursos Humanos.

---

**3. División de responsabilidades del equipo**

Cada integrante tendrá una responsabilidad principal, pero el desarrollo, la revisión de resultados, las pruebas y las decisiones de diseño serán actividades compartidas.

**Integrante 1: Gestión y preparación de datos** (Pier Herbozo)

* Desarrollar funciones reutilizables de carga y consulta para el resto del equipo.

* **Resultado esperado:** Estructura de datos limpia, documentada y lista para su procesamiento, que sirva como punto de partida para los cálculos de vacaciones y criticidad.

**Integrante 2: Análisis de vacaciones e impacto económico** (David Vargas)

* Definir reglas de negocio para clasificar vacaciones (pendientes, próximas a vencer y vencidas)
* Calcular días pendientes, salarios diarios y el costo económico asociado a cada escenario (incluida venta parcial de días)
* Identificar y priorizar a los empleados que requieren atención inmediata.

* **Resultado esperado:** Módulo de análisis financiero y operativo de vacaciones para la toma de decisiones

**Integrante 3: Análisis de criticidad de puestos y empleados** (Jorge Alarcón)

* **Resultado esperado:** Módulo de criticidad para identificar puestos y empleados relevantes para la continuidad operativa de la empresa
