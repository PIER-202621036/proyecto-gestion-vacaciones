## Depuración de la data

```python
# ==================================================================
# 3.-ZONA COMUN: Limpieza y validación (datos.py)
# ==================================================================

# Iteramos hoja por hoja
for nombre_hoja, df in negocio_excel.items():

    # 1. Duplicados
    cantidad_duplicados = df.duplicated().sum()
    if cantidad_duplicados > 0:
        print(f"Hoja {nombre_hoja}: {cantidad_duplicados} duplicados.")
        df = df.drop_duplicates()

    # 2. Nulos (excluyendo NULOS_PERMITIDOS)
    columnas_obligatorias = [
        col
        for col in df.columns
        if col.lower() not in [n.lower() for n in NULOS_PERMITIDOS]
    ]
    cantidad_nulos = df[columnas_obligatorias].isna().sum().sum()
    if cantidad_nulos > 0:
        print(f"Hoja {nombre_hoja}: {cantidad_nulos} nulos no permitidos.")
        df = df.dropna(subset=columnas_obligatorias)

    # Guardamos la hoja limpia de vuelta en el diccionario
    negocio_excel[nombre_hoja] = df.reset_index(drop=True)
```

## Operaciones de Negocio Comun

```python
# ==================================================================
# 4.- ZONA COMUN  (datos.py)
# Devuelve el empleado como diccionario, con el nombre de su puesto
# obtener_empleado(empleado_id, datos), para imdemnizacion y criticidad
# ==================================================================

# 1. ID del empleado en duro
empleado_id = 91

# 2. Convertir los DataFrames a listas de diccionarios sencillos
empleados = negocio_excel["Empleados"].to_dict(orient="records")
puestos = negocio_excel["Puestos"].to_dict(orient="records")

# 3. Bucle tradicional
resultado_obtener_empleado = None

for emp in empleados:
    if emp["id"] == empleado_id:
        resultado_obtener_empleado = emp.copy()
        for p in puestos:
            if (p["puesto_id"] == emp["puesto_id"]):
                resultado_obtener_empleado["puesto"] = p["puesto"]
                break
        break

# 4. Validar si lo encontró
print("\nInformación del empleado:")
print("=" * 60)
if resultado_obtener_empleado is None:
  print(f"No existe el empleado {empleado_id}")
else:
  #print(resultado_obtener_empleado)
  display(pd.DataFrame([resultado_obtener_empleado]))
```

```python
# ==================================================================
# 5.- ZONA COMUN  (datos.py)
# Devuelve los periodos del empleado, del mas antiguo al mas nuevo
# obtener_periodos(empleado_id, datos)
# ==================================================================

"""Devuelve los periodos del empleado, del mas antiguo al mas nuevo.
Es una lista de diccionarios. Si aun no tiene periodos, devuelve [].
"""

# Inicializamos la variable por defecto
resultado_periodos_empleado = []

# 1. Si resultado_obtener_empleado is None, no existe el empleado
if resultado_obtener_empleado is not None:

  vacaciones = negocio_excel["Vacaciones"].to_dict(orient="records")

  # 2. Filtrar las vacaciones del empleado usando un bucle for tradicional
  resultado_periodos_empleado = []

  for v in vacaciones:
      if v["empleado_id"] == empleado_id:
          resultado_periodos_empleado.append(v)

  # 3. Guardar y mostrar el resultado
  print(f"\nPeríodos de vacaciones encontrados para el empleado con id: {empleado_id}:")
  print("=" * 60)
  #print(pd.DataFrame(resultado_periodos_empleado).to_string(index=False))
  display(pd.DataFrame(resultado_periodos_empleado))

else:
    print(f"\nNo se encontraron períodos de vacaciones porque el empleado {empleado_id} no existe.")
```

## Módulo de Vencidas

```python
# ==================================================================
# 5.- ZONA COMUN  (datos.py)
# Devuelve el estado de los periodos (vencido, vigente )
# periodos_vencidos(periodos, fecha_corte=FECHA_CORTE)
# periodos_vigentes(periodos, fecha_corte=FECHA_CORTE)
# ==================================================================

# Devuelve solo los periodos que ya vencieron.
resultado_vencido_empleado = []
# Devuelve solo los periodos que esta vigente.
resultado_vigente_empleado = []

# 3. Recorrer la lista de períodos
for periodo in resultado_periodos_empleado:

    # Convertir la fecha fin del período con try/except inline
    try:
        texto_fecha_fp = str(periodo["fecha_fin_periodo"])[:10]
        fecha_fp = date.fromisoformat(texto_fecha_fp)
    except (ValueError, KeyError, TypeError):
        print(f"Fecha inválida o ausente: {periodo.get('fecha_fin_periodo')!r}")
        continue  # Salta al siguiente registro si la fecha falla

    # Evaluar si está Vencido o Vigente (comparando fechas)
    estado = ("Vencido" if fecha_fp < date.fromisoformat(FECHA_CORTE) else "Vigente")

    # Armar el diccionario reducido
    registro_limpio = {
        "empleado_id": periodo.get("empleado_id"),
        "vacacion_id": periodo.get("vacacion_id"),
        "periodo": periodo.get("periodo"),
        "fecha_fin_periodo": periodo.get("fecha_fin_periodo"),
        "dias_tomados": periodo.get("dias_tomados"),
        #"tomados_segun": periodo.get("tomados_segun"),
        "Estado": estado,
    }

    # Clasificar en la lista correspondiente
    if estado == "Vencido":
        resultado_vencido_empleado.append(registro_limpio)
    else:
        resultado_vigente_empleado.append(registro_limpio)

# 4. Mostrar el resultado de forma elegante
if len(resultado_vencido_empleado) > 0:
    print(f"\nPeríodos Vencidos (Fecha de corte: {FECHA_CORTE}):")
    print("=" * 60)
    display(pd.DataFrame(resultado_vencido_empleado))

if len(resultado_vigente_empleado) > 0:
    print(f"\nPeríodos Vigentes (Fecha de corte: {FECHA_CORTE}):")
    print("=" * 60)
    display(pd.DataFrame(resultado_vigente_empleado))

# Validación global: solo si AMBAS listas están vacías
if (
    len(resultado_vencido_empleado) == 0
    and len(resultado_vigente_empleado) == 0
):
    print(
        f"\nNo se encontraron períodos para el empleado a la fecha de corte ({FECHA_CORTE})."
    )
```

```python
# ==================================================================
# PASO 1: El empleado, la fecha de corte y las reglas de pago
# ==================================================================
# Se busca al empleado por su id y se leen las reglas de pago
# (descuento, sueldos por concepto, etc.) de la hoja Parametros_Vacaciones.

empleado_id = 91             # id del empleado a consultar,
fecha_corte = "2026-10-12"   # un período vence si su fecha de fin es anterior a esta fecha

# Textos que usan las hojas del Excel (se escriben tal cual estan)
TIPO_VENTA = "Venta_50"
RESPUESTA_ACEPTO = "Acepto"
ESTADO_PROGRAMADA = "Programada"
DECISION_APROBADA = "Aprobada"
TIPO_INDEMNIZACION = "Indemnizacion"
OFERTA_PAGADA = "Pagado"
OFERTA_AVISADA = "Avisado"

# Data del empleado
df_empleados = negocio_excel["Empleados"]
df_empleado = df_empleados[df_empleados["id"] == empleado_id]   # filtrado condicional: solo la fila de este empleado

print(f"\n\nMostrando al empleado: {empleado_id}")
print("="*140)
if df_empleado.empty:
    raise ValueError(f"No existe el empleado {empleado_id}")   # detiene la celda: sin empleado no hay boleta
display(df_empleado)
empleado = df_empleado.to_dict(orient="records")[0]   # el empleado como diccionario (para leer su salario)

# Las reglas de pago
df_parametros = negocio_excel["Parametros_Vacaciones"]

parametros = {}   # diccionario {nombre de la regla: valor}
for fila in df_parametros.to_dict(orient="records"):
    parametros[fila["regla"]] = int(fila["valor"])

print("\n\nReglas de pago (hoja Parametros_Vacaciones):")
print("="*140)
display(df_parametros[["regla", "valor", "descripcion"]])
```

```python
# ==================================================================
# PASO 2: Los períodos del empleado y si es indemnizable o no
# ==================================================================
# Días sin cubrir = días pendientes - días vendidos - días programados y aprobados
#   * vendidos:    ofertas Venta_50 que el empleado aceptó
#   * programados: bloques "Programada" con decisión "Aprobada" de RR. HH.
# Un período está Vencido si su fecha de fin es anterior a la fecha de corte.
# Es INDEMNIZABLE si está Vencido y todavía tiene días sin cubrir.

# Períodos del empleado, del más antiguo al más nuevo
df_periodos = negocio_excel["Vacaciones"]
periodos = df_periodos[df_periodos["empleado_id"] == empleado_id].to_dict(orient="records")
periodos.sort(key=lambda periodo: periodo["periodo"])   # lambda: ordena por el año del período

# Sus ofertas y sus bloques de vacaciones programadas
df_ofertas = negocio_excel["Ofertas_Compensacion"]
ofertas = df_ofertas[df_ofertas["empleado_id"] == empleado_id].to_dict(orient="records")

df_programacion = negocio_excel["Programacion_Vacaciones"]
programacion = df_programacion[df_programacion["empleado_id"] == empleado_id].to_dict(orient="records")

filas_periodos = []
for periodo in periodos:
    id_periodo = periodo["vacacion_id"]

    # Días vendidos de este período
    dias_vendidos = 0
    for oferta in ofertas:
        if (oferta["vacacion_id"] == id_periodo
                and oferta["tipo_oferta"] == TIPO_VENTA
                and oferta["estado"] == RESPUESTA_ACEPTO):
            dias_vendidos += oferta["dias"]

    # Días programados y aprobados de este período
    dias_programados = 0
    for bloque in programacion:
        if (bloque["vacacion_id"] == id_periodo
                and bloque["estado"] == ESTADO_PROGRAMADA
                and bloque["decision_rrhh"] == DECISION_APROBADA):
            dias_programados += bloque["dias"]

    # Lo que falta cubrir (nunca baja de 0)
    dias_sin_cubrir = max(0, periodo["dias_pendientes"] - dias_vendidos - dias_programados)

    # La fecha se pasa a texto AAAA-MM-DD, que se puede comparar directamente
    # str(...)[:10] deja la fecha como texto AAAA-MM-DD (quita la hora si la trae)
    # y así se puede comparar con fecha_corte. Ej: "2025-09-15 00:00:00" -> "2025-09-15"
    fecha_fin = str(periodo["fecha_fin_periodo"])[:10]
    if fecha_fin < fecha_corte:
        estado = "Vencido"
    else:
        estado = "Vigente"

    filas_periodos.append({
        "periodo": periodo["periodo"],
        "vacacion_id": id_periodo,
        "fecha_fin": fecha_fin,
        "dias_pendientes": periodo["dias_pendientes"],
        "dias_vendidos": dias_vendidos,
        "dias_programados": dias_programados,
        "dias_sin_cubrir": dias_sin_cubrir,
        "estado": estado,
        "indemnizable": estado == "Vencido" and dias_sin_cubrir > 0,
    })

print(f"\n\nPeríodos del empleado {empleado_id} (fecha de corte {fecha_corte}):")
print("="*140)
display(pd.DataFrame(filas_periodos))
```

```python
# ==================================================================
# PASO 3: HISTORIAL: los períodos indemnizables y cómo va su pago
# ==================================================================
# El historial muestra TODO lo que el empleado tiene, incluso lo ya pagado.
# La gestión de cada período indemnizable puede ser:
#   * "Pagada":       ya se le pagó -> NO entra en la boleta
#   * "Avisada":      RR. HH. ya le avisó, falta pagar
#   * "Falta avisar": todavía no se le avisa

historial = []
for fila in filas_periodos:
    if not fila["indemnizable"]:
        continue   # solo interesan los períodos vencidos con días sin cubrir

    gestion = "Falta avisar"
    for oferta in ofertas:
        if oferta["vacacion_id"] == fila["vacacion_id"] and oferta["tipo_oferta"] == TIPO_INDEMNIZACION:
            if oferta["estado"] == OFERTA_PAGADA:
                gestion = "Pagada"
            elif oferta["estado"] == OFERTA_AVISADA and gestion != "Pagada":
                gestion = "Avisada"   # Pagada tiene prioridad sobre Avisada

    historial.append({
        "periodo": fila["periodo"],
        "vacacion_id": fila["vacacion_id"],
        "dias_sin_cubrir": fila["dias_sin_cubrir"],
        "gestion": gestion,
    })

print(f"\n\nHistorial de indemnización del empleado {empleado_id}:")
print("="*140)
if historial:
    display(pd.DataFrame(historial))
else:
    print("Sin períodos indemnizables")
```

```python
# ==================================================================
# PASO 4: La BOLETA del próximo pago (lo que se le paga ahora)
# ==================================================================
# Orden de la boleta:
#   1. Sueldo del mes: el mes COMPLETO, UNA sola vez, con descuento.
#   2. Por cada período pendiente (del más antiguo al más nuevo):
#        - Remuneración vacacional = sueldo x factor, CON descuento
#        - Indemnización           = sueldo x factor, SIN descuento
#      donde factor = días sin cubrir / 30 (la parte del período que falta)
# Los períodos "Pagada" no entran: ya se pagaron.

salario = empleado["salario"]
descuento = parametros["porcentaje_descuento"]   # 20 (%)

lineas_boleta = []

# 1. Sueldo del mes (una sola vez, sin factor)
sueldo_bruto = round(salario * parametros["sueldos_mes_goce"], 2)
sueldo_descuento = round(sueldo_bruto * descuento / 100, 2)
lineas_boleta.append({
    "concepto": "Sueldo del mes",
    "periodo": "-",
    "dias": "-",
    "gestion": "-",
    "bruto": sueldo_bruto,
    "descuento": sueldo_descuento,
    "neto": round(sueldo_bruto - sueldo_descuento, 2),
})

# 2. Remuneración e indemnización de cada período pendiente
for periodo in historial:
    if periodo["gestion"] == "Pagada":
        continue   # ya se pagó: no forma parte de la boleta

    factor = periodo["dias_sin_cubrir"] / parametros["dias_vacaciones_periodo"]

    remuneracion = round(salario * parametros["sueldos_remuneracion_vacacional"] * factor, 2)
    remuneracion_descuento = round(remuneracion * descuento / 100, 2)
    indemnizacion = round(salario * parametros["sueldos_indemnizacion"] * factor, 2)

    lineas_boleta.append({
        "concepto": "Remuneración vacacional",
        "periodo": periodo["periodo"],
        "dias": periodo["dias_sin_cubrir"],
        "gestion": periodo["gestion"],
        "bruto": remuneracion,
        "descuento": remuneracion_descuento,
        "neto": round(remuneracion - remuneracion_descuento, 2),
    })
    lineas_boleta.append({
        "concepto": "Indemnización vacacional",
        "periodo": periodo["periodo"],
        "dias": periodo["dias_sin_cubrir"],
        "gestion": periodo["gestion"],
        "bruto": indemnizacion,
        "descuento": 0.0,   # la indemnización no lleva descuento
        "neto": indemnizacion,
    })

print(f"\n\nBoleta del próximo pago del empleado {empleado_id}:")
print("="*140)
display(pd.DataFrame(lineas_boleta))
```

```python
# ==================================================================
# PASO 5: Los totales de la boleta
# ==================================================================
# Un acumulador es una variable que empieza en 0 y va sumando dentro de un bucle.

total_bruto = 0
total_descuento = 0
total_neto = 0
for linea in lineas_boleta:
    total_bruto += linea["bruto"]
    total_descuento += linea["descuento"]
    total_neto += linea["neto"]

# Días de indemnización que se pagan en esta boleta (sin contar los ya pagados)
dias_a_pagar = 0
for periodo in historial:
    if periodo["gestion"] != "Pagada":
        dias_a_pagar += periodo["dias_sin_cubrir"]

print(f"\n\nTotales de la boleta del empleado {empleado_id}:")
print("="*140)
print(f"TOTAL -> bruto: S/ {total_bruto:,.2f} | descuento: S/ {total_descuento:,.2f} | neto: S/ {total_neto:,.2f}")
print(f"Días de indemnización a pagar: {dias_a_pagar}")




######################
# OTROS CASOS
######################

# empleado_id = 18   -> Pedro Cárdenas (sueldo 8,900): 2024 ya Pagada (no entra) y 2025 Avisada de 10 días
#                       3 líneas | bruto 14,833.34 | descuento 2,373.33 | neto 12,460.01 | 10 días

# empleado_id = 138  -> Jesús Herrera (sueldo 13,000): solo 5 días sin cubrir en 2024
#                       3 líneas | bruto 17,333.34 | descuento 3,033.33 | neto 14,300.01 | 5 días

# empleado_id = 1    -> Verónica Ramírez (sueldo 3,450): aún no tiene períodos
#                       solo el sueldo del mes | bruto 3,450.00 | descuento 690.00 | neto 2,760.00 | 0 días

# empleado_id = 99999 -> no existe: la celda 1 se detiene con "No existe el empleado 99999"
```

```python
# Buscar un puesto por su id y mostrar su detalle (incluye las 4 notas de criticidad)
puesto_id = "P09"

df_puestos = negocio_excel["Puestos"]
df_puesto = df_puestos[df_puestos["puesto_id"] == puesto_id]   # filtrado condicional

print(f"\n\nMostrando detalles del Puesto: {puesto_id}")
print("="*140)
if df_puesto.empty:
    print(f"No existe el puesto {puesto_id}")
else:
    display(df_puesto)
```

```python
# Cada variable se califica del 1 al 5,
# donde el 1 es lo menos grave y el 5 tiene mayor impacto.

df_reglas = negocio_excel["Parametros_Criticidad"]

print("\n\nReglas de criticidad (cada variable se califica con una nota):")
print("="*140)
display(df_reglas[["variable", "peso", "nota_min", "nota_max", "nota_1_significa", "nota_5_significa"]])
```

```python
# Cálculo del score
# Para P09 la tabla debe dar 40, 20, 25 y 6 puntos, que suman 91.
# Cada fila sale de (nota / 5 × peso × 100);
# por ejemplo, la de dificultad_reemplazo es 4 / 5 × 0.25 × 100 = 20.

puesto = df_puesto.to_dict(orient="records")[0]      # el puesto como diccionario
reglas = df_reglas.to_dict(orient="records")         # cada regla como diccionario

filas_calculo = []
score = 0

for regla in reglas:
    nombre = regla["variable"]
    nota = puesto[nombre]                            # nota del puesto en esta variable

    if nota < regla["nota_min"] or nota > regla["nota_max"]:
        print(f"ATENCION: la nota de {nombre} ({nota}) esta fuera de {regla['nota_min']} a {regla['nota_max']}")

    puntos = nota / regla["nota_max"] * regla["peso"] * 100
    score += puntos

    filas_calculo.append({
        "variable": nombre,
        "nota": nota,
        "nota_max": regla["nota_max"],
        "peso": regla["peso"],
        "puntos": round(puntos, 2),
    })

score = round(score, 2)

print(f"\n\nCalculo del score del Puesto: {puesto_id}")
print("="*140)
display(pd.DataFrame(filas_calculo))
print(f"SCORE = {score}")
```

```python
#El resultado es Score 91.0 -> Nivel Critico

df_niveles = excel_hojas["Niveles_Criticidad"]
niveles = df_niveles.to_dict(orient="records")
niveles.sort(key=lambda nivel: nivel["score_min"], reverse=True)   # del nivel mas alto al mas bajo

nivel_puesto = None
for nivel in niveles:
    if score >= nivel["score_min"]:
        nivel_puesto = nivel["nivel"]
        break

print(f"\n\nNivel de criticidad del Puesto: {puesto_id}")
print("="*140)
display(df_niveles)
print(f"Score {score} -> Nivel {nivel_puesto}")


######################
#OTROS CASOS
######################

# Para probar otro puesto,
# se cambia solo el puesto_id en la celda 1
# y se vuelve a correr las cuatra celdas.
# puesto_id = P12 sale 20.0 y Bajo,
# y con puesto_id = P03 sale 100 y Critico.
```
