```python
# ************************************************************
# CARGAR EXCEL DESDE GITHUB
# Configuración de la fuente de datos
# ************************************************************


import pandas as pd

#Ruta del excel en GitHub
URL_EXCEL = (
    "https://raw.githubusercontent.com/"
    "PIER-202621036/proyecto-gestion-vacaciones/"
    "main/data/empleados_vacaciones.xlsx"
)

# Cargar todas las hojas del Excel PIER
excel_hojas = pd.read_excel(URL_EXCEL, sheet_name=None)

# Mostrar las hojas disponibles
print("Hojas cargadas:")
for nombre_hoja in excel_hojas:
    print("*", nombre_hoja)

#Mostrar la tabla directamente display sirve para eso
excel_hoja_empleados= excel_hojas["Empleados"]
print("Hoja Empleados")
display(excel_hoja_empleados) # excel_hoja_vacaciones = excel_hojas["Vacaciones"]
```

```python
# ************************************************************
# SEMANA 1 — VARIABLES, TIPOS Y CONSTANTES
# ************************************************************

import pandas as pd

#CONSTANTES DEL PROYECTO
# 1. Convertimos la hoja de Excel a una lista de registros/filas
filas = excel_hojas["Parametros_Vacaciones"].to_dict("records")

# 2. Creamos un diccionario vacío
parametros = {}

# 3. Llenamos el diccionario fila por fila
for fila in filas:
    nombre_regla = fila["regla"]
    valor_numero = int(fila["valor"])

    parametros[nombre_regla] = valor_numero

# Constantes del negocio
DIAS_VACACIONES_PERIODO = parametros["dias_vacaciones_periodo"]
PORCENTAJE_VENTA = parametros["porcentaje_venta"]
PERIODOS_CRITICOS = parametros["periodos_criticos"]
PORCENTAJE_DESCUENTO = parametros["porcentaje_descuento"]
DIAS_ALERTA_VENCIMIENTO = parametros["dias_alerta_vencimiento"]

# Fecha de corte definida en el Plan_S1_S7
FECHA_CORTE = "2026-10-12"

constantes = {
    "DIAS_VACACIONES_PERIODO": DIAS_VACACIONES_PERIODO,
    "PORCENTAJE_VENTA": PORCENTAJE_VENTA,
    "PERIODOS_CRITICOS": PERIODOS_CRITICOS,
    "PORCENTAJE_DESCUENTO": PORCENTAJE_DESCUENTO,
    "DIAS_ALERTA_VENCIMIENTO": DIAS_ALERTA_VENCIMIENTO,
    "FECHA_CORTE": FECHA_CORTE,
}

print("CONSTANTES DEL PROYECTO")
print("=" * len("CONSTANTES DEL PROYECTO"))
for nombre, valor in constantes.items():
    print(f"{nombre} = {valor!r} | Tipo: {type(valor).__name__}")

#DICCIONARIO DE TIPOS DE DATOS
excel_hojas_principales = [
    "Empleados",
    "Puestos",
    "Vacaciones",
    "Programacion_Vacaciones",
    "Ofertas_Compensacion",
]

tipos_campos = []

for fila in excel_hojas["Diccionario"].to_dict("records"):
    nombre_hoja = fila["hoja"]

    # Si la hoja no está en nuestra lista principal, la evitamos
    if nombre_hoja not in excel_hojas_principales:
        continue

    campo = fila["campo"]

    # Valida si el nombre de la hoja que sacamos de [Diccionario] no existe en el excel
    # O si el campo no existe en la hoja del excel
    if (nombre_hoja not in excel_hojas or campo not in excel_hojas[nombre_hoja].columns):
        tipo = "Campo no encontrado"
    else:
        valores = excel_hojas[nombre_hoja][campo].dropna() # dropna Elimina todos los elementos vacíos.

        # Valida si han quedado datos
        if valores.empty:
            tipo = "Sin datos"
        else:
            valor = valores.iloc[0] #obtiene el 1er elemento encontrado

            # Convertir tipos de NumPy a tipos nativos de Python para leer su nombre
            if hasattr(valor, "item"):
                valor = valor.item()
            tipo = type(valor).__name__

    tipos_campos.append(
        {
            "hoja": nombre_hoja,
            "campo": campo,
            "tipo_python_observado": tipo,
        }
    )

print("\n\nHOJA DICCIONARIO Y TIPOS DE DATOS")
print("=" * len("DICCIONARIO DE TIPOS DE DATOS"))
display(pd.DataFrame(tipos_campos))


```

```python

# ============================================================
# SEMANA 2 — CONDICIONALES Y BUCLES
# validación de integridad de datos para la lista de empleados
# ============================================================

mensaje_errores = []

muestra_empleados = excel_hojas["Empleados"].head(10).to_dict("records")
lista_puestos = excel_hojas["Puestos"].to_dict("records")

# 1. Revisar salario y fecha de ingreso de 10 empleados
for empleado in muestra_empleados:
    puesto = None

    # asigna puesto si existe
    for p in lista_puestos:
        if p["puesto_id"] == empleado["puesto_id"]:
            puesto = p
            break

    if puesto is None:
        mensaje_errores.append({
            "tabla": "Empleados",
            "id": empleado["id"],
            "error": "El puesto_id no existe"
        })
    # valida si salario está dentro del rango permitido
    elif (
        empleado["salario"] < puesto["banda_min"]
        or empleado["salario"] > puesto["banda_max"]
    ):
        mensaje_errores.append({
            "tabla": "Empleados",
            "id": empleado["id"],
            "error": "Salario fuera de la banda del puesto"
        })

    #valida si es un valor nulo/vacío (notna)
    fecha_ingreso = empleado["fecha_ingreso"]
    if pd.notna(fecha_ingreso) and str(fecha_ingreso) > FECHA_CORTE:
        mensaje_errores.append({
            "tabla": "Empleados",
            "id": empleado["id"],
            "error": "Fecha de ingreso posterior a FECHA_CORTE"
        })

print("\nRESULTADO DE LAS VALIDACIONES")
print("=" * len("RESULTADO DE LAS VALIDACIONES"))

if mensaje_errores:
    display(pd.DataFrame(mensaje_errores))
else:
    print("No se encontraron errores en las muestras revisadas.")
```
