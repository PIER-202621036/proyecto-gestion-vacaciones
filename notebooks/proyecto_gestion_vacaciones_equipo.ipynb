# **II. Guía de estilo del proyecto PEP 8**

## **PEP 8** es la guía oficial de estilo de Python y **PEP 257** es la guía
oficial de docstrings. No cambian lo que hace el programa, solo cómo se ve
y cómo se nombra. Como somos tres personas escribiendo funciones que luego
se unen, todos seguimos estas reglas para que el código parezca escrito por
una sola persona.

Fuentes (consultadas el 14-oct-2026):

- PEP 8, guía de estilo: <https://peps.python.org/pep-0008/>
- PEP 257, convenciones de docstrings: <https://peps.python.org/pep-0257/>

PEP 8 dice que, si un proyecto tiene su propia guía, esa guía manda para
ese proyecto. Este documento es la guía de **nuestro** proyecto: toma lo
esencial de los PEP y fija las decisiones del equipo.

---

## Regla 1. Nombres en `snake_case`

Variables y funciones van en minúsculas, separando palabras con guion bajo.
(PEP 8, sección *Function and Variable Names*.)

**Mal**
```python
SalarioMensual = 10800
def calcularIndemnizacion(salario, diasPendientes):
    ...
```

**Bien**
```python
salario_mensual = 10800
def calcular_indemnizacion(salario, dias_pendientes):
    ...
```

---

## Regla 2. Constantes en MAYÚSCULAS

Los valores que no cambian (las reglas de la hoja `Reglas_Vacaciones`) se
escriben en mayúsculas, con guion bajo, al inicio del archivo.
(PEP 8, sección *Constants*.)

**Mal**
```python
descuento = 20
dias = 30
```

**Bien**
```python
PORCENTAJE_DESCUENTO = 20
DIAS_VACACIONES_PERIODO = 30
PERIODOS_CRITICOS = 2
```

---

## Regla 3. Docstring en cada función

Un texto entre triples comillas dobles, como **primera línea dentro** de la
función. Python lo guarda en `__doc__`, y por eso `help(funcion)` y Colab lo
muestran. (PEP 257 y PEP 8, sección *Documentation Strings*.)

Condiciones del PEP 257:

- Siempre triples comillas dobles `"""`.
- Es una frase que termina en punto y se escribe como **orden**
  ("Devuelve...", "Calcula..."), no como descripción
  ("Esta función devuelve...").
- Las funciones simples llevan **una sola línea**, con las comillas de
  cierre en la misma línea.
- Las funciones del contrato (las que usan los otros dos) pueden llevar
  más: resumen, línea en blanco, y qué recibe y qué devuelve.

**Mal**
```python
def obtener_empleado(empleado_id):
    ...
```

**Bien (función simple)**
```python
def obtener_empleado(empleado_id):
    """Devuelve los datos del empleado como diccionario."""
    ...
```

**Bien (función del contrato)**
```python
def obtener_periodos(empleado_id):
    """Devuelve la lista de períodos vacacionales del empleado.

    Recibe el id del empleado (int).
    Devuelve una lista de diccionarios, uno por período.
    """
    ...
```

> El `...` solo marca "aquí va el código". No se escribe en el programa real.

---

## Regla 4. Líneas cortas y espacios alrededor de los operadores

PEP 8 (sección *Maximum Line Length*) fija **79 caracteres** por línea y
**72** para comentarios y docstrings. También dice que un equipo puede
acordar un límite mayor, hasta 99. **Nuestro acuerdo: 79.**

Si no cabe, se parte dentro de paréntesis. Un espacio a cada lado de `=`,
`+`, `-`, `*`, `/` cuando se pueda (PEP 8, *Whitespace in Expressions and
Statements*). En argumentos con nombre no se pone espacio: `f(a, b=2)`.

**Mal**
```python
total=calcular_indemnizacion(salario_mensual,dias_pendientes)["bruto"]+calcular_venta_50(salario_mensual,dias_vendibles)["bruto"]
```

**Bien**
```python
indemnizacion = calcular_indemnizacion(salario_mensual, dias_pendientes)
venta = calcular_venta_50(salario_mensual, dias_vendibles)
total = indemnizacion["bruto"] + venta["bruto"]
```

---

## Regla 5. Comentarios que explican el porqué

Se comenta la razón de una decisión, no lo que ya dice el código.
Los comentarios son frases completas, empiezan con mayúscula y terminan en
punto. Un comentario en la misma línea lleva dos espacios antes del `#`.
(PEP 8, sección *Comments*.)

**Mal**
```python
factor = dias_pendientes / 30   # divide entre 30
```

**Bien**
```python
# Se paga en proporción si quedan menos de 30 días pendientes.
factor = dias_pendientes / DIAS_VACACIONES_PERIODO
```

Idioma: PEP 8 recomienda comentarios en inglés, salvo que el código solo
vaya a ser leído por personas de tu idioma. **Nuestro acuerdo: comentarios y
docstrings en español.** Nombres de variables y funciones también en español,
sin tildes ni eñes (`antiguedad`, no `antigüedad`).

---

## Ejemplo completo con todas las reglas

```python
DIAS_VACACIONES_PERIODO = 30
PORCENTAJE_DESCUENTO = 20


def calcular_indemnizacion(salario, dias_pendientes):
    """Calcula la indemnización de un período vencido.

    Recibe el salario mensual (float) y los días pendientes (int).
    Devuelve un diccionario {'bruto': float, 'neto': float}.
    """
    factor = dias_pendientes / DIAS_VACACIONES_PERIODO
    descuento = PORCENTAJE_DESCUENTO / 100

    # El primer sueldo va íntegro; los otros dos llevan descuento.
    bruto = salario * 3 * factor
    neto = salario * factor * (1 + 2 * (1 - descuento))
    return {"bruto": bruto, "neto": neto}


# Prueba: período con 15 días pendientes y sueldo de 10,800.
resultado = calcular_indemnizacion(10800, 15)
print(resultado)  # {'bruto': 16200.0, 'neto': 14040.0}
```

Este resultado coincide con la hoja `Ej4_Venta50` del Excel (período 2024
del empleado 91).
